# Generate Random Resampling Test Data

Generates random load profiles, resamples them with both
the resampling method, and saves the results as
CSV/JSON files for use in downstream tests.

Define multiple cases in the `CASES` list below. Each case is a dict
with the generation parameters.

## Imports and output directory

```python
import datetime
import os

from ethos_penalps.testing.load_profile.load_profil_creator import make_random_load_profile_entries
from ethos_penalps.testing.load_profile.resampling_helpers import (
    resample_load_profile,
    save_resampling_test_case,
)

CASES_DIR = os.path.join(os.path.dirname(os.path.abspath("__file__")), "cases")
```

## Case definitions

Each entry defines one random test case. Add or remove entries as needed.

```python
CASES = [
    {
        "start": datetime.datetime(2023, 1, 1, 0, 0, 0),
        "end": datetime.datetime(2023, 1, 1, 0, 30, 0),
        "n_entries": 15,
        "max_energy": 1000.0,
        "seed": 1042,
        "resample_frequency": "7.5min",
    },
    {
        "start": datetime.datetime(2021, 1, 1, 0, 0, 0),
        "end": datetime.datetime(2021, 1, 1, 0, 15, 0),
        "n_entries": 50,
        "max_energy": 500.0,
        "seed": 7,
        "resample_frequency": "1.25min",
    },
]
```

## Generate, resample and save all cases

```python
for i, case in enumerate(CASES):
    start = case["start"]
    end = case["end"]
    n_entries = case["n_entries"]
    max_energy = case["max_energy"]
    seed = case["seed"]
    resample_frequency = case["resample_frequency"]

    total_duration_s = int((end - start).total_seconds())
    case_name = f"random_seed_{seed}_{total_duration_s}s_{resample_frequency}"

    print(f"--- Case {i + 1}/{len(CASES)}: {case_name} ---")

    entries = make_random_load_profile_entries(start, end, n_entries, max_energy, seed)
    print(f"  Generated {len(entries)} input entries over {end - start}")

    meta, result = resample_load_profile(entries, start, end, resample_frequency)

    print(f"  Input entries:       {len(meta.list_of_load_profiles)}")
    print(f"  Output bins:         {len(result.list_of_load_profiles)}")
    print(f"  Input total energy:  {meta.total_energy:.6f} {meta.energy_unit}")
    print(f"  Output total energy: {result.total_energy:.6f} {result.energy_unit}")

    parameters = {
        "start": start.isoformat(),
        "end": end.isoformat(),
        "n_entries": n_entries,
        "max_energy": max_energy,
        "seed": seed,
        "resample_frequency": resample_frequency,
    }

    save_resampling_test_case(
        meta=meta,
        result=result,
        cases_dir=CASES_DIR,
        parameters=parameters,
        case_name=case_name,
    )
    print()
```
