# Storage Entry Visualization

Loads pre-generated test cases from any subfolder and visualizes:
1. Input streams, storage level, and output streams (gantt chart)
2. Cumulative mass with and without offset

Select a folder and case using the dropdowns below.

```python
import os

%load_ext autoreload
%autoreload 2

from ethos_penalps.testing.storage.storage_test_case_io import (
    list_available_cases,
    load_storage_test_case,
)
from ethos_penalps.testing.storage.storage_test_visualization import (
    plot_cumulative_mass,
    plot_streams_and_storage,
)

BASE_DIR = os.path.dirname(os.path.abspath("__file__"))
```

## Select Folder and Case

```python
FOLDERS = [
    "01_single_continuous_stream",
    "02_single_batch_stream",
    "03_balanced_timing",
    "04_multiple_streams",
    "05_mixed_stream_types",
    "06_time_bounds",
]

# --- Choose folder and case ---
FOLDER = FOLDERS[5]
CASE_NAME = "multi_stream_cut"  # Set to a specific case name, or None for the first available

cases_dir = os.path.join(BASE_DIR, FOLDER, "cases")
available_cases = list_available_cases(cases_dir)
print(f"Folder: {FOLDER}")
print(f"Available cases:")
for c in available_cases:
    print(f"  - {c}")

if CASE_NAME is None and available_cases:
    CASE_NAME = available_cases[0]
print(f"\nSelected case: {CASE_NAME}")
```

## Load Case

```python
case_dir = os.path.join(cases_dir, CASE_NAME)
storage_test_case = load_storage_test_case(case_dir)
storage_test_case.print_detailed_summary()
```

## 1. Input Streams, Storage Level, Output Streams

```python
plot_streams_and_storage(storage_test_case);
```

## 2. Cumulative Mass (with and without offset)

```python
plot_cumulative_mass(storage_test_case)
```
