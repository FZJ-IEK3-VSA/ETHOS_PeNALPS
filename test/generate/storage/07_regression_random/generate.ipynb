# Regression Random (50 states)

Six seeded random test cases with 50 input and 50 output stream states each,
covering all combinations of continuous and batch streams.

| Case | Input | Output |
|------|-------|--------|
| 1 | Continuous 1x50 | Continuous 1x50 |
| 2 | Continuous 1x50 | Batch 1x50 |
| 3 | Batch 1x50 | Continuous 1x50 |
| 4 | Batch 1x50 | Batch 1x50 |
| 5 | Continuous 5x10 | Continuous 5x10 |
| 6 | Continuous 5x10 | Batch 5x10 |

```python
import datetime
import os
import random

%load_ext autoreload
%autoreload 2

from ethos_penalps.testing.storage.storage_test_case import (
    StorageTestCaseSpecification,
    build_case_from_parameters,
)
from ethos_penalps.testing.storage.storage_test_case_io import save_storage_test_case
from ethos_penalps.testing.stream.stream_group_test_case import (
    StreamGroupTestCaseSpecification,
)
from ethos_penalps.testing.stream.stream_test_case import (
    BatchStreamStateParams as B,
    BatchStreamTestCaseSpecification,
    ContinuousStreamStateParams as C,
    ContinuousStreamTestCaseSpecification,
)

CASES_DIR = os.path.join(os.path.dirname(os.path.abspath("__file__")), "cases")
T0 = datetime.datetime(2024, 1, 1)
SEED = 42
```

```python
def random_continuous_states(rng: random.Random, n: int) -> list[C]:
    """Generate *n* random continuous stream state parameters."""
    states = []
    for _ in range(n):
        mass = rng.uniform(5, 100)
        rate = rng.uniform(1, 50)
        duration = mass / rate * 3600  # seconds
        gap = rng.uniform(0, 1800)
        states.append(C(mass=mass, rate=rate, duration_seconds=duration, gap_after_seconds=gap))
    return states


def random_batch_states(rng: random.Random, n: int) -> list[B]:
    """Generate *n* random batch stream state parameters."""
    states = []
    for _ in range(n):
        mass = rng.uniform(5, 100)
        gap = rng.uniform(600, 3600)
        states.append(B(mass=mass, gap_seconds=gap))
    return states
```

## Continuous input / Continuous output

```python
rng = random.Random(SEED + 1)

params = StorageTestCaseSpecification(
    name="cont_in_cont_out",
    description="50 continuous input / 50 continuous output states.",
    start_time=T0,
    start_storage_level="auto_offset",
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=random_continuous_states(rng, 50),
            )
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=random_continuous_states(rng, 50),
                start_offset_seconds=1800,
            )
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```

## Multi-stream Continuous input / Continuous output

```python
rng = random.Random(SEED + 5)

params = StorageTestCaseSpecification(
    name="multi_cont_in_cont_out",
    description="5x10 continuous input / 5x10 continuous output states.",
    start_time=T0,
    start_storage_level="auto_offset",
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=random_continuous_states(rng, 10),
                start_offset_seconds=i * 7200,
            )
            for i in range(5)
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=random_continuous_states(rng, 10),
                start_offset_seconds=900 + i * 7200,
            )
            for i in range(5)
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```
