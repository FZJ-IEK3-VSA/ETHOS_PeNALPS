# Balanced Timing

Mass-conserved cases (input = output) with 1 continuous input and 1 continuous output.
Varies: timing order × storage level (level_0, auto_offset).

`simultaneous` has no offset variant since storage stays flat at 0.

```python
import datetime
import os

%load_ext autoreload
%autoreload 2

from ethos_penalps.stream import StreamType
from ethos_penalps.testing.storage.storage_test_case import (
    StorageTestCaseSpecification,
    build_case_from_parameters,
)
from ethos_penalps.testing.storage.storage_test_case_io import save_storage_test_case
from ethos_penalps.testing.stream.stream_group_test_case import (
    StreamGroupTestCaseSpecification,
)
from ethos_penalps.testing.stream.stream_test_case import (
    BatchStreamStateParams as B,
)
from ethos_penalps.testing.stream.stream_test_case import (
    BatchStreamTestCaseSpecification,
    ContinuousStreamTestCaseSpecification,
)
from ethos_penalps.testing.stream.stream_test_case import (
    ContinuousStreamStateParams as C,
)

CASES_DIR = os.path.join(os.path.dirname(os.path.abspath("__file__")), "cases")
T0 = datetime.datetime(2024, 1, 1)
```

## Simultaneous — balanced

Input and output at same rate and time. Storage stays flat at zero.

```python
params = StorageTestCaseSpecification(
    name="simultaneous_continuous",
    description="Input and output at same rate and time, storage stays flat at zero.",
    start_time=T0,
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=100, rate=10, duration_seconds=3600)])
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=100, rate=10, duration_seconds=3600)])
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```

## Input before output — level_0

Input fills 100 in hour 1, output drains 100 in hour 2. start_storage_level=0.

```python
params = StorageTestCaseSpecification(
    name="input_before_output_level_0",
    description="Input fills 100 in hour 1, output drains 100 in hour 2. start_storage_level=0.",
    start_time=T0,
    start_storage_level=0,
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=100, rate=10, duration_seconds=3600)]
            ),
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=100, rate=10, duration_seconds=3600)], start_offset_seconds=3600
            ),
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```
