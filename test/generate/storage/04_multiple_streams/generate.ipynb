# Multiple Streams

Mass-conserved cases with multiple input and/or output streams.
All use auto_offset. Varies: fan-in/fan-out topology, stream type (continuous/batch).

```python
import datetime
import os

%load_ext autoreload
%autoreload 2

from ethos_penalps.stream import StreamType
from ethos_penalps.testing.storage.storage_test_case import (
    StorageTestCaseSpecification,
    build_case_from_parameters,
)
from ethos_penalps.testing.storage.storage_test_case_io import save_storage_test_case
from ethos_penalps.testing.stream.stream_group_test_case import (
    StreamGroupTestCaseSpecification,
)
from ethos_penalps.testing.stream.stream_test_case import (
    BatchStreamStateParams as B,
)
from ethos_penalps.testing.stream.stream_test_case import (
    BatchStreamTestCaseSpecification,
    ContinuousStreamTestCaseSpecification,
)
from ethos_penalps.testing.stream.stream_test_case import (
    ContinuousStreamStateParams as C,
)

CASES_DIR = os.path.join(os.path.dirname(os.path.abspath("__file__")), "cases")
T0 = datetime.datetime(2024, 1, 1)
```

## Two continuous inputs, one continuous output

Two upstream processes at different rates (10/h and 5/h) feeding a single downstream consumer.

```python
params = StorageTestCaseSpecification(
    name="two_continuous_inputs_one_continuous_output",
    description="Two continuous input streams at different rates feeding one continuous output.",
    start_time=T0,
    start_storage_level="auto_offset",
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=100, rate=10, duration_seconds=3600)]),
            ContinuousStreamTestCaseSpecification(states=[C(mass=100, rate=5, duration_seconds=7200)]),
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=200, rate=10, duration_seconds=7200)]),
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```

## One continuous input, two continuous outputs

A single upstream process feeds two downstream consumers that draw at different times.

```python
params = StorageTestCaseSpecification(
    name="one_continuous_input_two_continuous_outputs",
    description="One continuous input stream feeding two continuous output streams at different times.",
    start_time=T0,
    start_storage_level="auto_offset",
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=200, rate=10, duration_seconds=7200)]),
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=80, rate=10, duration_seconds=3600)]),
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=120, rate=10, duration_seconds=3600)], start_offset_seconds=7200
            ),
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```

## Two continuous inputs, two continuous outputs — staggered

Two continuous input streams and two continuous output streams with staggered timing. Storage must buffer mass between all four streams.

```python
params = StorageTestCaseSpecification(
    name="two_continuous_inputs_two_continuous_outputs_staggered",
    description="Two continuous inputs and two continuous outputs with staggered timing.",
    start_time=T0,
    start_storage_level="auto_offset",
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(states=[C(mass=100, rate=10, duration_seconds=3600)]),
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=100, rate=10, duration_seconds=3600)], start_offset_seconds=7200
            ),
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=80, rate=10, duration_seconds=3600)], start_offset_seconds=3600
            ),
            ContinuousStreamTestCaseSpecification(
                states=[C(mass=120, rate=10, duration_seconds=3600)], start_offset_seconds=10800
            ),
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```

## Two batch inputs, one batch output

Two batch input streams at different offsets feeding a single batch output stream.

```python
params = StorageTestCaseSpecification(
    name="two_batch_inputs_one_batch_output",
    description="Two batch input streams at different offsets feeding one batch output.",
    start_time=T0,
    start_storage_level="auto_offset",
    expect_full_mass_conservation=True,
    input_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            BatchStreamTestCaseSpecification(
                states=[B(mass=60, gap_seconds=7200), B(mass=40, gap_seconds=7200)],
                start_offset_seconds=3600,
                batch_delay_seconds=3600,
                max_batch_mass=100,
            ),
            BatchStreamTestCaseSpecification(
                states=[B(mass=50, gap_seconds=7200), B(mass=50, gap_seconds=7200)],
                start_offset_seconds=7200,
                batch_delay_seconds=3600,
                max_batch_mass=100,
            ),
        ]
    ),
    output_stream_group=StreamGroupTestCaseSpecification(
        stream_specifications=[
            BatchStreamTestCaseSpecification(
                states=[B(mass=100, gap_seconds=3600), B(mass=100, gap_seconds=3600)],
                start_offset_seconds=14400,
                batch_delay_seconds=3600,
                max_batch_mass=200,
            ),
        ]
    ),
)

case = build_case_from_parameters(params)
case.print_overview()
save_storage_test_case(case, CASES_DIR)
```
